### Data to be cleaned


```python
import numpy as np 
import pandas as pd
```

```python
data = pd.read_csv(r'..\Datasets\credit_risk_dataset.csv')
```

```python
data.shape[0]
```

```python
data.drop_duplicates(inplace=True)
data.shape[0]
```

```python
# Valid age 
data = data[data['person_age']>=18]
data = data[data['person_age']<=100]
```

```python
#Remove employement length grater then age and the number 60
data = data[data['person_emp_length'] < data['person_age']]
data = data[data['person_emp_length'] <=60]
```

```python

data=data[data['loan_amnt']>0]
```

```python
data.to_csv('../Datasets/credit_risk_dataset_cleaned.csv')
```
